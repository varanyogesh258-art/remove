 ## **MATRIX PROGRAM**

```python
row=int(input("enter the no of row :"))
col=int(input("enter the no of column :"))
symbol=input("enter the symbol :")
i=1
while i<=row:
    j=1
    while j<=col:
        print(symbol,end="")
        j=j+1
    i=i+1
    print()
```

## **TRIANGLE PROGRAM**


```python
row=int(input("enter the no of row :"))
symbol=input("enter the symbol :")
i=1
while i<=row:
    j=1
    while j<=i:
        print(symbol,end="")
        j=j+1
    i=i+1
    print()
```
