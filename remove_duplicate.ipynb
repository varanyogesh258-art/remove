## **REMOVE DUPLICATE**

```python
def remove_duplicates(mylist):
    newlist = []
    for each in mylist:
        if each not in newlist:
            newlist+=[each]
    return newlist  
inputlist=[20,30.55,"hai","hello",20,100,30.55,True,"hello","hai",100,False]
print(remove_duplicates(inputlist))
```
